```python
import psycopg2

conn = psycopg2.connect(
    host="postgres",
    database="frauddb",
    user="frauduser",
    password="fraudpass"
)

print("Connected!")
```

```python
cur = conn.cursor()

cur.execute("""
CREATE TABLE IF NOT EXISTS processed_transactions (
    transaction_id        VARCHAR(50) PRIMARY KEY,
    user_id               VARCHAR(20),
    merchant_id           VARCHAR(20),
    amount                NUMERIC(12,2),
    currency              VARCHAR(10),
    merchant_category     VARCHAR(50),
    payment_method        VARCHAR(50),
    device                VARCHAR(50),
    user_country          VARCHAR(50),
    transaction_country   VARCHAR(50),
    transaction_timestamp TIMESTAMP,
    fraud_score           INTEGER,
    risk_level            VARCHAR(20),
    fraud_reasons         TEXT,
    is_fraud_alert        BOOLEAN
)
""")

conn.commit()
```

```python
cur = conn.cursor()

cur.execute("""
CREATE TABLE IF NOT EXISTS fraud_alerts (
    alert_id              BIGSERIAL PRIMARY KEY,
    transaction_id        VARCHAR(50),
    user_id               VARCHAR(20),
    amount                NUMERIC(12,2),
    fraud_score           INTEGER,
    risk_level            VARCHAR(20),
    fraud_reasons         TEXT,
    transaction_timestamp TIMESTAMP,
    alert_created_at      TIMESTAMP DEFAULT CURRENT_TIMESTAMP
)
""")

conn.commit()
```
